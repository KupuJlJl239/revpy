```python
from revpy import reversible, ReversibleFunction, reversibility_assert


def add_inplace(none, x, a):
    reversibility_assert(none is None, none)
    x[0] += a

def sub_inplace(none, x, a):
    reversibility_assert(none is None, none)
    x[0] -= a

add_inplace = ReversibleFunction(add_inplace, sub_inplace)
sub_inplace = add_inplace.rev

a = 10
def make_f():
    x = [1]
    n = 3

    @reversible
    def f(none):
        reversibility_assert(none is None, none)
        for i in range(n):
            add_inplace(x=x, a=a*i)[None]
        return none

    def print_x():
        print(x[0])

    return f, print_x

f, print_x = make_f()

print_x()
f[None]
print_x()
f[None]
print_x()
f.rev[None]
print_x()
```
